```python
import os
import numpy as np
import h5py
from skimage.metrics import structural_similarity as ssim
```

```python
output_dir = '/mnt/newdisk/diffusion/code_zjy/project/diffusion/DDPM/zjy/DDPM/output/test/result_epoch_174'
mae_list = []
ssim_list = []
for file in os.listdir(output_dir):
    if file.endswith('.hdf5'):
        with h5py.File(os.path.join(output_dir, file), 'r') as f:
            pred = f['output'][()]
            gt = f['target'][()]
            input = f['input'][()]
            #print(f"{file}, Pred shape: {pred.shape}, GT shape: {gt.shape}, Input shape: {input.shape}")
            mae = np.mean(np.abs(pred - gt))
            ssim_value = ssim(pred, gt, data_range=1.0, channel_axis=None)
            print(f"MAE: {mae}, SSIM: {ssim_value}")
            mae_list.append(mae)
            ssim_list.append(ssim_value)

print(f"avg MAE: {np.mean(mae_list)}, avg SSIM: {np.mean(ssim_list)}")
```
